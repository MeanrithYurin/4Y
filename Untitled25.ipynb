```python
# -- Sheet --

import pandas as pd
import numpy as np
import datetime as dt
import os
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', 500)

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv('/content/sample_data/BNK.csv')
```

```python
df["Payment_Due_Date"] = pd.to_datetime(
    df["Payment_Due_Date"].astype(str).str.split('.').str[0]
)
df["Month"] = df["Payment_Due_Date"].dt.strftime("%Y%m")
```

```python
group_cols = [
    "ReportDate",
    "Arrangement_ID",
    "Loan_Account_Number",
    "Product_ID",
    "Product_Name",
    "Currency",
    "Maturity_Date",
    "Interest_Rate",
    "Loan_Term",
    "Outstanding_Balance_USD",
    "Accured_Interest_USD",
    "Outstanding_Balance_Orig"
]
```

```python
segment_map = {
    # Agriculture Loan
    "Agriculture Loan": "Agriculture Loan",
    "Power Tiller": "Agriculture Loan",

    # Business Loan
    "Individual Credit- Business Loan": "Business Loan",

    # Consumption Loan
    "Consumption Loan": "Consumption Loan",
    "Fix Deposit Loan": "Consumption Loan",
    "Smart EDU Loan": "Consumption Loan",

    # Home Improvement Loan
    "Home Improvement Loan": "Home Improvement Loan",
    "Small Housing Loan": "Home Improvement Loan",

    # SME Loan
    "SME Loan": "SME Loan",
    "SME loan": "SME Loan",
    "SME Co Funding Scheme": "SME Loan",
    "Short Term Loan": "SME Loan",
    "Fast loan": "SME Loan",
    "3 Wheels": "SME Loan",
    "Trade Loan": "SME Loan",
    "Office Construction Loan": "SME Loan",

    # Solidarity Credit Loan
    "Solidarity Credit": "Solidarity Credit Loan",
    "Non-Collateral Loan": "Solidarity Credit Loan",
    "INSTALMENTLOAN": "Solidarity Credit Loan",

    # Staff Loan
    "Personal Asset Staff Loan": "Staff Loan",
    "Vehicle Staff Loan": "Staff Loan"
}

df["Segment"] = df["Product_Name"].map(segment_map)
```

```python
group_cols = [
    "ReportDate",
    "Arrangement_ID",
    "Loan_Account_Number",
    "Product_ID",
    "Product_Name",
    "Segment",
    "Currency",
    "Maturity_Date",
    "Interest_Rate",
    "Loan_Term",
    "Outstanding_Balance_USD",
    "Accured_Interest_USD",
    "Outstanding_Balance_Orig"
]
```

```python
result = pd.pivot_table(
    df,
    index=group_cols,
    columns="Month",
    values="Installment_Principle_Orig",
    aggfunc="sum",
    fill_value=0
)
result = result.reset_index()
```

```python
# result["MIA"] = ""
# result["Stage"] = ""

# result["VALUE_HARD_TITLE"] = ""
# result["VALUE_SOFT_TITLE"] = ""
# result["VALUE_DEPOSIT"] = ""
# result["VALUE_MOVEABLE_ID"] = ""
# result["VALUE_OTHER"] = ""
```

```python
report_month = (
    pd.to_datetime(
        result["ReportDate"].astype(str),
        format="%Y%m%d"
    )
    .dt.strftime("%Y%m")
    .min()
)

max_month = df["Month"].dropna().max()

months = pd.period_range(
    start=pd.Period(report_month, freq="M"),
    end=pd.Period(max_month, freq="M"),
    freq="M"
)

months = [
    str(m).replace("-", "")
    for m in months
]
```

```python
for m in months:
    if m not in result.columns:
        result[m] = 0
```

```python
final_cols = [
    "Arrangement_ID",
    "Loan_Account_Number",
    "Product_ID",
    "Product_Name",
    "Segment",
    "Currency",
    "Maturity_Date",
    "Interest_Rate",
    "Loan_Term",
    "Number_of_Days_Late",
    "ReportDate",
    "Disbursed_Amount_USD",
    "Outstanding_Balance_USD",
    "Accured_Interest_USD",
    "Outstanding_Balance_Orig",
] + months
```

```python

cols_to_select = [col for col in final_cols if col in result.columns]
result = result[cols_to_select]
```

```python
output_file = "Loan_EAD_Output_FinalBNK.xlsx"

result.to_excel(
    output_file,
    index=False
)
```
